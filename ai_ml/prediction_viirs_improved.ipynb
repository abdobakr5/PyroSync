# VIIRS Fire Radiative Power (FRP) prediction â€” improved pipeline

**Target:** `frp` (heavy-tailed, always >= 0)

What changed compared with the original notebook:

| Area | Original | Now |
|---|---|---|
| Model | XGBoost on `log1p(FRP)`, squared error | **LightGBM with a Tweedie objective** (log-link, predicts the conditional *mean* on the raw scale) + a weighted log-target LightGBM, **blended** on validation |
| Tuning metric | log-RMSE, but reported in raw RMSE (mismatch) | Tweedie model tuned directly on **raw RMSE**; blend weight tuned on raw RMSE |
| Features | raw columns only (`acq_time` dropped) | + hour of day, cyclic time features, pixel area, TI4-TI5 difference/ratio, same-day neighbourhood fire statistics, per-cell history and out-of-fold cell target encoding |
| Back-transform bias | `expm1(pred)` systematically under-predicts large fires | handled by Tweedie model and the blend |
| Evaluation | RMSE / MAE / R2 | + MedAE, RMSLE, R2 in log space, **error by FRP range**, plots, feature importance |
| Code hygiene | loader code copied 3x, duplicated cells | one loader function, one feature function applied identically to every split |

## 2. Feature engineering

The same function is applied to every split. Everything here uses only the **row's own columns and
the other detections on the same day within the same split**, so there is no target leakage.

```python
DROP_COLS = ["frp","instrument", "acq_time", "version", "acq_date"]

def make_features(df):
    d = df.copy()
    d["acq_date"] = pd.to_datetime(d["acq_date"], errors="coerce")

    # --- time ---
    t = pd.to_numeric(d["acq_time"], errors="coerce").fillna(0).astype(int)
    d["hour"] = t // 100 + (t % 100) / 60
    d["hour_sin"] = np.sin(2 * np.pi * d["hour"] / 24)
    d["hour_cos"] = np.cos(2 * np.pi * d["hour"] / 24)
    doy = d["acq_date"].dt.dayofyear
    d["day_of_year"] = doy
    d["doy_sin"] = np.sin(2 * np.pi * doy / 365.25)
    d["doy_cos"] = np.cos(2 * np.pi * doy / 365.25)
    d["month"] = d["acq_date"].dt.month

    # --- sensor / physics ---
    d["pixel_area"] = d["scan"] * d["track"]
    d["ti_diff"] = d["bright_ti4"] - d["bright_ti5"]
    denom = d["bright_ti5"].replace(0, np.nan)
    d["ti_ratio"] = d["bright_ti4"] / denom

    # --- space ---
    lat_bin = np.floor((d["latitude"] + 90) / CELL_SIZE_DEG).astype(int)
    lon_bin = np.floor((d["longitude"] + 180) / CELL_SIZE_DEG).astype(int)
    d["cell_id"] = lat_bin * 10000 + lon_bin
    d["ti4_saturated"] = (
    d["bright_ti4"] >= 367
).astype("int8")
    d["is_day"] = (
    d["daynight"] == "D"
).astype("int8")

    # --- same-day neighbourhood (fire cluster size / intensity) ---
    g = d.groupby(["acq_date", "cell_id"])["bright_ti4"]
    # d["day_cell_n"] = g.transform("size")
    # d["day_cell_ti4_mean"] = g.transform("mean")
    # d["day_cell_ti4_max"] = g.transform("max")
    # d["ti4_vs_day_cell_mean"] = d["bright_ti4"] - d["day_cell_ti4_mean"]

    d["_lat1"] = np.floor(d["latitude"])
    d["_lon1"] = np.floor(d["longitude"])
    #d["day_1deg_n"] = d.groupby(["acq_date", "_lat1", "_lon1"])["bright_ti4"].transform("size")

    # --- how "fire-prone" is this cell in this year (unsupervised) ---
    # d["year_cell_n"] = d.groupby(["year", "cell_id"])["cell_id"].transform("size")

    return d.drop(columns=DROP_COLS + ["_lat1", "_lon1"], errors="ignore")


X_tr0 = make_features(train_raw);  y_tr = train_raw["frp"].to_numpy()
X_va0 = make_features(val_raw);    y_va = val_raw["frp"].to_numpy()
X_te0 = make_features(test_raw);   y_te = test_raw["frp"].to_numpy()

# categoricals: fix categories on train so every split has identical encodings
cat_cols = X_tr0.select_dtypes(include=["object", "string", "category"]).columns.tolist()
for c in cat_cols:
    cats = X_tr0[c].dropna().unique()
    for X in (X_tr0, X_va0, X_te0):
        X[c] = pd.Categorical(X[c], categories=cats)

print("categorical:", cat_cols)
print("columns match:", list(X_tr0.columns) == list(X_va0.columns) == list(X_te0.columns))
print(X_tr0.shape, X_va0.shape, X_te0.shape)
print("year_cell_n" in X_tr0.columns)
print("Infinite values:", np.isinf(X_tr0.select_dtypes(include=np.number)).sum().sum())
print("Missing ti_ratio:", X_tr0["ti_ratio"].isna().sum())
```

### Cell target encoding (out-of-fold)

Average `log1p(FRP)` of the grid cell, smoothed towards the global mean. For the training rows it is
computed **out-of-fold** so a row never sees its own target.

```python
def add_time_aware_cell_te(
    train_df,
    y_train,
    *future_dfs,
    cell_col="cell_id",
    year_col="year",
    smoothing=20
):
    """
    Training rows: encode each year using only earlier years.
    Future rows: encode using all years in train_df.
    Target encoding is calculated on log1p(FRP).
    """

    train_df = train_df.copy()
    future_dfs = [df.copy() for df in future_dfs]

    y_log = np.log1p(np.asarray(y_train))
    years = train_df[year_col].to_numpy()

    train_df["cell_te"] = np.nan
    train_df["cell_te_count"] = 0

    # 1. Chronological encoding for training rows
    for year in sorted(pd.unique(years)):

        current_mask = years == year
        history_mask = years < year

        # No earlier years available
        if not history_mask.any():
            continue

        history = pd.DataFrame({
            cell_col: train_df.loc[history_mask, cell_col].to_numpy(),
            "target": y_log[history_mask]
        })

        prior_mean = history["target"].mean()

        cell_stats = history.groupby(cell_col)["target"].agg(
            ["mean", "count"]
        )

        smoothed_encoding = (
            cell_stats["mean"] * cell_stats["count"]
            + smoothing * prior_mean
        ) / (cell_stats["count"] + smoothing)

        current_cells = train_df.loc[current_mask, cell_col]

        train_df.loc[current_mask, "cell_te"] = (
            current_cells.map(smoothed_encoding).fillna(prior_mean)
        )

        train_df.loc[current_mask, "cell_te_count"] = (
            current_cells.map(cell_stats["count"]).fillna(0)
        )

    # 2. Build encoding from all training years for future data
    full_history = pd.DataFrame({
        cell_col: train_df[cell_col].to_numpy(),
        "target": y_log
    })

    prior_mean = full_history["target"].mean()

    cell_stats = full_history.groupby(cell_col)["target"].agg(
        ["mean", "count"]
    )

    smoothed_encoding = (
        cell_stats["mean"] * cell_stats["count"]
        + smoothing * prior_mean
    ) / (cell_stats["count"] + smoothing)

    # 3. Apply the historical encoding to validation/test
    for df in future_dfs:

        df["cell_te"] = (
            df[cell_col].map(smoothed_encoding).fillna(prior_mean)
        )

        df["cell_te_count"] = (
            df[cell_col].map(cell_stats["count"]).fillna(0)
        )

    return (train_df, *future_dfs)


def drop_id(X):
    return X.drop(columns=["cell_id", "year"], errors="ignore")

# Stage 1: time-aware encoding using TRAIN only

X_tr, X_va, X_te = add_time_aware_cell_te(
    X_tr0,
    y_tr,
    X_va0,
    X_te0
)

# Drop cell_id
X_tr = drop_id(X_tr)
X_va = drop_id(X_va)
X_te = drop_id(X_te)
X_tr = X_tr.drop(columns=['satellite','daynight','pixel_area','hour_sin','day_of_year','ti_ratio','doy_sin'], errors="ignore")
X_va = X_va.drop(columns=['satellite','daynight','pixel_area','hour_sin','day_of_year','ti_ratio','doy_sin'], errors="ignore")
X_te = X_te.drop(columns=['satellite','daynight','pixel_area','hour_sin','day_of_year','ti_ratio','doy_sin'], errors="ignore")

FEATURES = X_tr.columns.tolist()

print(len(FEATURES), "features:", FEATURES)
```

## 3. Evaluation helpers

```python
def metrics(y, p, name):
    p = np.maximum(p, 0)
    return {
        "model": name,
        "RMSE": float(np.sqrt(mean_squared_error(y, p))),
        "MAE": float(mean_absolute_error(y, p)),
        "MedAE": float(median_absolute_error(y, p)),
        "R2": float(r2_score(y, p)),
        "RMSLE": float(np.sqrt(mean_squared_error(np.log1p(y), np.log1p(p)))),
        "R2_log": float(r2_score(np.log1p(y), np.log1p(p))),
    }

def error_by_range(y, p, edges=(0, 1, 5, 20, 100, np.inf)):
    p = np.maximum(p, 0)
    b = pd.cut(y, bins=list(edges), include_lowest=True)
    df = pd.DataFrame({"bin": b, "abs_err": np.abs(y - p), "bias": p - y})
    return df.groupby("bin", observed=True).agg(
        n=("abs_err", "size"), MAE=("abs_err", "mean"), bias=("bias", "mean"))

def make_weights(y):
    # mild emphasis on large fires, bounded so a few extreme rows cannot dominate
    w = 1.0 + np.log1p(y)
    return w / w.mean()
```

## 4. Model A â€” LightGBM with Tweedie objective (raw scale)

FRP is non-negative and right-skewed, which is exactly what the Tweedie family (variance power between 1 and 2)
models. With a log link, the model predicts the conditional **mean** of FRP directly, so there is no
`expm1` back-transform bias and it optimises what raw RMSE actually measures.

```python
def fit_lgb(
    params, X, y, X_val=None, y_val=None, w=None,
    metric="rmse", rounds=4000, categorical_features=None
):
    if not isinstance(X, pd.DataFrame):
        raise TypeError("X must be a pandas DataFrame.")

    X_train = X.copy()

    if categorical_features is None:
        cat_cols = X_train.select_dtypes(
            include=["object", "string", "category"]
        ).columns.tolist()
    else:
        cat_cols = list(categorical_features)

    # Validate categorical columns
    missing = set(cat_cols) - set(X_train.columns)
    if missing:
        raise ValueError(f"Missing categorical columns: {missing}")

    for col in cat_cols:
        X_train[col] = X_train[col].astype("category")

    m = lgb.LGBMRegressor(
        n_estimators=rounds,
        random_state=SEED,
        n_jobs=-1,
        verbose=-1,
        **params
    )

    fit_kwargs = {
        "sample_weight": w,
        "categorical_feature": cat_cols
    }

    if X_val is not None and y_val is not None:
        X_valid = X_val.copy()

        for col in cat_cols:
            X_valid[col] = pd.Categorical(
                X_valid[col],
                categories=X_train[col].cat.categories
            )

        fit_kwargs.update({
            "eval_set": [(X_valid, y_val)],
            "eval_metric": metric,
            "callbacks": [
                lgb.early_stopping(
                    100,
                    first_metric_only=True,
                    verbose=False
                )
            ]
        })

    m.fit(X_train, y, **fit_kwargs)

    return m



def objective_tweedie(trial):
    params = {
        "objective": "tweedie",
        "metric": "rmse",
        "tweedie_variance_power": trial.suggest_float("tweedie_variance_power", 1.05, 1.9),
        "learning_rate": trial.suggest_float("learning_rate", 0.02, 0.15, log=True),
        "num_leaves": trial.suggest_int("num_leaves", 31, 255, log=True),
        "min_child_samples": trial.suggest_int("min_child_samples", 10, 200, log=True),
        "feature_fraction": trial.suggest_float("feature_fraction", 0.6, 1.0),
        "bagging_fraction": trial.suggest_float("bagging_fraction", 0.6, 1.0),
        "bagging_freq": 1,
        "lambda_l1": trial.suggest_float("lambda_l1", 1e-3, 10, log=True),
        "lambda_l2": trial.suggest_float("lambda_l2", 1e-3, 10, log=True),
    }
    m = fit_lgb(params, X_tr, y_tr, X_va, y_va)
    trial.set_user_attr("best_iter", int(m.best_iteration_))
    return float(np.sqrt(mean_squared_error(y_va, np.maximum(m.predict(X_va), 0))))

study_a = optuna.create_study(direction="minimize", sampler=optuna.samplers.TPESampler(seed=SEED))
study_a.optimize(objective_tweedie, n_trials=N_TRIALS_TWEEDIE, show_progress_bar=True)
print("Best val RMSE (raw):", study_a.best_value)
print(study_a.best_params)
```

## 5. Model B â€” LightGBM on `log1p(FRP)` with sample weights

This one is good at the *typical* fire (median behaviour), while Model A is good at the *large* ones.
The blend takes the best of both.

```python
y_tr_log, y_va_log = np.log1p(y_tr), np.log1p(y_va)
w_tr = make_weights(y_tr) if WEIGHT_LOG_MODEL else None

def objective_log(trial):
    params = {
        "objective": "regression",
        "metric": "rmse",
        "learning_rate": trial.suggest_float("learning_rate", 0.02, 0.15, log=True),
        "num_leaves": trial.suggest_int("num_leaves", 31, 255, log=True),
        "min_child_samples": trial.suggest_int("min_child_samples", 10, 200, log=True),
        "feature_fraction": trial.suggest_float("feature_fraction", 0.6, 1.0),
        "bagging_fraction": trial.suggest_float("bagging_fraction", 0.6, 1.0),
        "bagging_freq": 1,
        "lambda_l1": trial.suggest_float("lambda_l1", 1e-3, 10, log=True),
        "lambda_l2": trial.suggest_float("lambda_l2", 1e-3, 10, log=True),
    }
    m = fit_lgb(params, X_tr, y_tr_log, X_va, y_va_log, w=w_tr)
    trial.set_user_attr("best_iter", int(m.best_iteration_))
    return float(m.best_score_["valid_0"]["rmse"])

study_b = optuna.create_study(direction="minimize", sampler=optuna.samplers.TPESampler(seed=SEED))
study_b.optimize(objective_log, n_trials=N_TRIALS_LOG, show_progress_bar=True)
print("Best val log-RMSE:", study_b.best_value)
print(study_b.best_params)
```

## 6. Blend weights (chosen on validation)

```python
params_a = {"objective": "tweedie", "metric": "rmse", **study_a.best_params}
params_b = {"objective": "regression", "metric": "rmse", **study_b.best_params}

model_a = fit_lgb(params_a, X_tr, y_tr, X_va, y_va)
model_b = fit_lgb(params_b, X_tr, y_tr_log, X_va, y_va_log, w=w_tr)

pa_val = np.maximum(model_a.predict(X_va), 0)
pb_val = np.maximum(np.expm1(model_b.predict(X_va)), 0)

grid = np.linspace(0, 1, 21)
rmses = [np.sqrt(mean_squared_error(y_va, a * pa_val + (1 - a) * pb_val)) for a in grid]
BLEND_A = float(grid[int(np.argmin(rmses))])
print(f"Blend weight on Tweedie model: {BLEND_A:.2f}")

val_table = pd.DataFrame([
    metrics(y_va, pa_val, "A: Tweedie"),
    metrics(y_va, pb_val, "B: log-target"),
    metrics(y_va, BLEND_A * pa_val + (1 - BLEND_A) * pb_val, "Blend"),
]).set_index("model")
val_table.round(4)
```

## 7. Final fit on train + validation, evaluate once on test

Cell encoding is recomputed on train+val, tree counts come from early stopping (x1.15 because there is more data).
The test set is touched only here.

```python
# rebuild features with TE computed from train+val
# Rebuild features using TRAIN + VALIDATION
X_dev0 = pd.concat([X_tr0, X_va0], ignore_index=True)
y_dev = np.concatenate([y_tr, y_va])

# Recalculate time-aware encoding using all development data
X_dev, X_te_f = add_time_aware_cell_te(
    X_dev0,
    y_dev,
    X_te0
)

# Remove identifier columns and select the same features
X_dev = drop_id(X_dev)[FEATURES]
X_te_f = drop_id(X_te_f)[FEATURES]

n_a = int(model_a.best_iteration_ * 1.15)
n_b = int(model_b.best_iteration_ * 1.15)
w_dev = make_weights(y_dev) if WEIGHT_LOG_MODEL else None

final_a = fit_lgb(params_a, X_dev, y_dev, rounds=n_a)
final_b = fit_lgb(params_b, X_dev, np.log1p(y_dev), w=w_dev, rounds=n_b)

pa_te = np.maximum(final_a.predict(X_te_f), 0)
pb_te = np.maximum(np.expm1(final_b.predict(X_te_f)), 0)
p_blend = BLEND_A * pa_te + (1 - BLEND_A) * pb_te

test_table = pd.DataFrame([
    {"model": "Original XGBoost-log (reported)", "RMSE": 14.13, "MAE": 3.63},
    metrics(y_te, pa_te, "A: Tweedie"),
    metrics(y_te, pb_te, "B: log-target"),
    metrics(y_te, p_blend, "Blend (final)"),
]).set_index("model")
test_table.round(4)
```

## Notes and next steps

* **Is `bright_ti4` / `bright_ti5` available at prediction time?** FRP is derived from the I4 channel radiance, so these
  columns explain most of the variance. If the real goal is to *forecast* FRP before observing a fire, remove
  `bright_ti4, bright_ti5, ti_diff, ti_ratio, ti4_vs_day_cell_mean, day_cell_ti4_*` from the features
  and expect a much lower score.
* Extreme FRP values dominate RMSE. Read RMSE together with MAE, MedAE, RMSLE and the per-range table above.
* If `type` = 2 (static sources) is not of interest, filter those rows out before training.
* Try `CatBoostRegressor(loss_function="Tweedie:variance_power=1.5")` or a quantile model (alpha 0.1/0.9)
  to get prediction intervals.
